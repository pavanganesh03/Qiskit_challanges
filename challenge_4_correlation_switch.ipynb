In this notebook I design a circuit that switches between correlated outcomes (00 and 11) and opposite outcomes (01 and 10).

## Step 1: Importing the libraries

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

sim = AerSimulator()
shots = 100
```

## Part A: Correlated outcomes
Circuit: H on qubit 0, CNOT from qubit 0 to qubit 1, then measure both. That is 2 gates.

**Prediction:** `00` and `11` will dominate with about 50 shots each. `01` and `10` should not appear, so not all four outcomes will show up.

```python
qc_a = QuantumCircuit(2, 2)
qc_a.h(0)
qc_a.cx(0, 1)
qc_a.measure([0, 1], [0, 1])
qc_a.draw('mpl')
```

```python
counts_a = sim.run(transpile(qc_a, sim), shots=shots).result().get_counts()
print("Part A results:", counts_a)
plot_histogram(counts_a, title="Part A: correlated outcomes")
```

## Part B: Opposite outcomes
I use the same circuit and add an X gate on qubit 1 after the CNOT. That is 3 gates in total.

**Prediction:** `01` and `10` will dominate with about 50 shots each, and `00` and `11` should not appear.

```python
qc_b = QuantumCircuit(2, 2)
qc_b.h(0)
qc_b.cx(0, 1)
qc_b.x(1)
qc_b.measure([0, 1], [0, 1])
qc_b.draw('mpl')
```

```python
counts_b = sim.run(transpile(qc_b, sim), shots=shots).result().get_counts()
print("Part B results:", counts_b)
plot_histogram(counts_b, title="Part B: opposite outcomes")
```

## Part C: Comparing both circuits

```python
plot_histogram([counts_a, counts_b], legend=['Part A', 'Part B'], title="Part A vs Part B")
```

```python
def percent(counts):
    return {k: counts.get(k, 0) / shots * 100 for k in ['00', '01', '10', '11']}

print("Part A (%):", percent(counts_a))
print("Part B (%):", percent(counts_b))
```

## Answers

**Which outcomes are dominant in Part A?** `00` and `11`, each close to 50%.

**Which outcomes are dominant in Part B?** `01` and `10`, each close to 50%.

**What changed between the two circuits?** Only one extra X gate on qubit 1 was added after the CNOT in Part B.

**Why did the measurement relationship change?** The H and CNOT gates entangle the qubits in the state (|00> + |11>)/sqrt(2), so both qubits always match. The X gate flips qubit 1 in every case, which turns |00> into |01> and |11> into |10>. The state becomes (|01> + |10>)/sqrt(2), so the qubits now always differ.

**Will all four outcomes appear?** No. In each circuit only two of the four outcomes appear. The other two are impossible because of the entanglement.

**Why are some outcomes much more frequent?** The amplitudes of two of the states are zero, so those outcomes have probability 0. The other two each have probability 1/2, so each is seen about 50 times in 100 shots, with a small random difference.

**How the circuit switches the correlation:** The H + CNOT part creates the entanglement, and the optional X gate on qubit 1 decides whether the result is "same" (00, 11) or "opposite" (01, 10).
