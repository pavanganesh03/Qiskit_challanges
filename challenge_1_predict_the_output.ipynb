In this notebook I build three single-qubit circuits (A, B and C), predict what they will give, run each one for 100 shots and then compare my predictions with the real results.

## Step 1: Importing the libraries

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

sim = AerSimulator()
shots = 100
```

## Step 2: Circuit A - one X gate
The qubit starts in |0>. The X gate flips |0> to |1>.

**My prediction:** I will get `1` every time, so all 100 shots should be `1`.

```python
qc_a = QuantumCircuit(1, 1)
qc_a.x(0)
qc_a.measure(0, 0)
qc_a.draw('mpl')
```

```python
job_a = sim.run(transpile(qc_a, sim), shots=shots)
counts_a = job_a.result().get_counts()
print("Circuit A results:", counts_a)
plot_histogram(counts_a, title="Circuit A: X gate")
```

## Step 3: Circuit B - two X gates
The first X makes |0> into |1> and the second X flips it back to |0>.

**My prediction:** I will get `0` every time, so all 100 shots should be `0`.

```python
qc_b = QuantumCircuit(1, 1)
qc_b.x(0)
qc_b.x(0)
qc_b.measure(0, 0)
qc_b.draw('mpl')
```

```python
job_b = sim.run(transpile(qc_b, sim), shots=shots)
counts_b = job_b.result().get_counts()
print("Circuit B results:", counts_b)
plot_histogram(counts_b, title="Circuit B: two X gates")
```

## Step 4: Circuit C - one H gate
The Hadamard gate puts the qubit in an equal superposition of |0> and |1>.

**My prediction:** I should get about 50 zeros and 50 ones. The result will not be exactly the same every time I run it, because measuring a superposition is random.

```python
qc_c = QuantumCircuit(1, 1)
qc_c.h(0)
qc_c.measure(0, 0)
qc_c.draw('mpl')
```

```python
job_c = sim.run(transpile(qc_c, sim), shots=shots)
counts_c = job_c.result().get_counts()
print("Circuit C results:", counts_c)
plot_histogram(counts_c, title="Circuit C: H gate")
```

## Step 5: All three histograms together

```python
plot_histogram([counts_a, counts_b, counts_c], legend=['A (X)', 'B (X, X)', 'C (H)'], title="Circuits A, B and C")
```

```python
print("Circuit A:", counts_a)
print("Circuit B:", counts_b)
print("Circuit C:", counts_c)
print()
print("Circuit C running again 3 times:")
for i in range(3):
    again = sim.run(transpile(qc_c, sim), shots=shots).result().get_counts()
    print(again)
```

## Step 6: Comparing my predictions with the results

| Circuit | Prediction | Actual result |
|---|---|---|
| A (one X) | all `1` | all 100 shots gave `1` |
| B (two X) | all `0` | all 100 shots gave `0` |
| C (H) | about 50% `0` and 50% `1` | close to 50-50, but the exact numbers change on every run |

**Conclusion:** My predictions for A and B were correct because the X gate is a definite flip, so these circuits are not random. In circuit C the H gate creates a superposition, so the answer is random and circuit C does **not** always give the same result. Its counts stay near 50-50 but are never exactly the same each time.
